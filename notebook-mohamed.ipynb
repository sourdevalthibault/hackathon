```python
y_train_path = "./sia-predicting-short-form-video-popularity/y_train.csv"
y_train = pd.read_csv(y_train_path, sep=";")
```

```python
df = X_train.merge(y_train, left_on="id", right_on="ID", how="inner")
```

```python
# df = X_train.merge(y_train, left_on="id", right_on="ID", how="inner")

target_col = "popularity"  # adapte si ta cible a un autre nom

num_df = df.select_dtypes(include="number").drop(columns=["id", "ID"], errors="ignore")

corr_matrix = num_df.corr(numeric_only=True)
corr_with_target = corr_matrix[target_col].sort_values(ascending=False)

corr_matrix, corr_with_target
```
