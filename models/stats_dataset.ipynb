```python
import pandas as pd
import numpy as np
X_train = pd.read_csv("../features/global_features/X_train.csv")
y_train = pd.read_csv("../features/global_features/y_train.csv")
X_test = pd.read_csv("../features/global_features/X_test.csv")
SEUIL = 0.15 #0.069
SEUIL_clip = 0.1 #0.14
```

```python
X_train.head()
```

```python
X_train_video = X_train.iloc[:,np.r_[:23, 535:547, 609:632]]
X_train_video.head()
```

```python
X_train_clip = X_train.iloc[:,23:535]
X_train_clip.head()
```

```python
X_train_audio = X_train.iloc[:,547:586]
X_train_audio.head()
```

```python
X_train_meta = X_train.iloc[:,586:609]
X_train_meta.head()
```

```python
X_train_vigg = X_train.iloc[:,632:]
X_train_vigg.head()
```

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# 1. Toujours centrer et réduire (scaler) avant une PCA
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_train_clip)

# 2. On instancie la PCA sans fixer le nombre de composantes pour tout voir
pca = PCA(n_components=0.95)
X_pca = pca.fit_transform(X_scaled)

# 3. Conversion en DataFrame Pandas
# On crée des noms de colonnes dynamiques : PC1, PC2, PC3...
column_names = [f'c_pc{i+1}' for i in range(X_pca.shape[1])]
X_train_pca_clip = pd.DataFrame(data=X_pca, columns=column_names, index=X_train_clip.index)

# --- Affichage des résultats ---
print(f"Nombre de features initiales : {X_train_clip.shape[1]}")
print(f"Nombre de composantes retenues : {pca.n_components_}")
print(f"Variance totale expliquée : {pca.explained_variance_ratio_.sum():.2%}")

# Affichage des premières lignes du nouveau DataFrame
print(X_train_pca_clip.head())
```

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# 1. Toujours centrer et réduire (scaler) avant une PCA
scaler_vigg = StandardScaler()
X_scaled = scaler_vigg.fit_transform(X_train_vigg)

# 2. On instancie la PCA sans fixer le nombre de composantes pour tout voir
pca_vigg = PCA(n_components=0.95)
X_pca = pca_vigg.fit_transform(X_scaled)

# 3. Conversion en DataFrame Pandas
# On crée des noms de colonnes dynamiques : PC1, PC2, PC3...
column_names = [f'v_pc{i+1}' for i in range(X_pca.shape[1])]
X_train_pca_vigg = pd.DataFrame(data=X_pca, columns=column_names, index=X_train_vigg.index)

# --- Affichage des résultats ---
print(f"Nombre de features initiales : {X_train_vigg.shape[1]}")
print(f"Nombre de composantes retenues : {pca_vigg.n_components_}")
print(f"Variance totale expliquée : {pca_vigg.explained_variance_ratio_.sum():.2%}")

# Affichage des premières lignes du nouveau DataFrame
print(X_train_pca_vigg.head())
```

```python
full_df = pd.concat([X_train_pca_clip, y_train], axis=1)
matrix = full_df.corr()
matrix = matrix.abs()
target_corr = matrix['popularity'].sort_values(ascending=False)
print(target_corr.head(20))
```

```python
full_df = pd.concat([X_train_pca_clip, y_train], axis=1)
matrix = full_df.corr().abs()
target_corr = matrix['popularity'][matrix['popularity'] > SEUIL_clip]
target_corr = target_corr.drop(labels=['popularity'], errors='ignore')
features_to_keep = target_corr.index.tolist()
X_train_clip_filtered = X_train_pca_clip[features_to_keep]

# --- Affichage ---
print(f"Nombre de features avant : {X_train_pca_clip.shape[1]}")
print(f"Nombre de features après filtrage : {len(features_to_keep)}")
print(f"Variables conservées : {features_to_keep}")
```

```python
full_df = pd.concat([X_train_pca_vigg, y_train], axis=1)
matrix = full_df.corr().abs()
target_corr = matrix['popularity'][matrix['popularity'] > SEUIL_clip]
target_corr = target_corr.drop(labels=['popularity'], errors='ignore')
features_to_keep = target_corr.index.tolist()
X_train_vigg_filtered = X_train_pca_vigg[features_to_keep]

# --- Affichage ---
print(f"Nombre de features avant : {X_train_pca_vigg.shape[1]}")
print(f"Nombre de features après filtrage : {len(features_to_keep)}")
print(f"Variables conservées : {features_to_keep}")
```

```python
full_df = pd.concat([X_train_video, y_train], axis=1)
matrix = full_df.corr().abs()
target_corr = matrix['popularity'][matrix['popularity'] > SEUIL]
target_corr = target_corr.drop(labels=['popularity'], errors='ignore')
features_to_keep = target_corr.index.tolist()
X_train_video_filtered = X_train_video[features_to_keep]

# --- Affichage ---
print(f"Nombre de features avant : {X_train_video.shape[1]}")
print(f"Nombre de features après filtrage : {len(features_to_keep)}")
print(f"Variables conservées : {features_to_keep}")
```

```python
full_df = pd.concat([X_train_meta, y_train], axis=1)
matrix = full_df.corr().abs()
target_corr = matrix['popularity'][matrix['popularity'] > SEUIL]
target_corr = target_corr.drop(labels=['popularity'], errors='ignore')
features_to_keep = target_corr.index.tolist()
X_train_meta_filtered = X_train_meta[features_to_keep]

# --- Affichage ---
print(f"Nombre de features avant : {X_train_meta.shape[1]}")
print(f"Nombre de features après filtrage : {len(features_to_keep)}")
print(f"Variables conservées : {features_to_keep}")
```

```python
full_df = pd.concat([X_train_audio, y_train], axis=1)
matrix = full_df.corr().abs()
target_corr = matrix['popularity'][matrix['popularity'] > SEUIL]
target_corr = target_corr.drop(labels=['popularity'], errors='ignore')
features_to_keep = target_corr.index.tolist()
X_train_audio_filtered = X_train_audio[features_to_keep]

# --- Affichage ---
print(f"Nombre de features avant : {X_train_audio.shape[1]}")
print(f"Nombre de features après filtrage : {len(features_to_keep)}")
print(f"Variables conservées : {features_to_keep}")
```

```python
X_train_filtered = pd.concat([X_train.iloc[:,0],X_train_audio_filtered,X_train_clip_filtered,X_train_meta_filtered,X_train_video_filtered,X_train_vigg_filtered], axis=1)
# --- Affichage ---
print(f"Nombre de features après : {X_train_filtered.shape[1]}")
```

```python
# 1. Découpage des blocs du X_test (même logique que pour X_train)
X_test_video_raw = X_test.iloc[:, np.r_[:23, 535:547, 609:632]]
X_test_clip_raw = X_test.iloc[:, 23:535]
X_test_audio_raw = X_test.iloc[:, 547:586]
X_test_meta_raw = X_test.iloc[:, 586:609]
X_test_vigg_raw = X_test.iloc[:, 632:]

# 2. Transformation PCA du bloc 'clip'
# Note: On utilise le scaler et la pca du train (cellule 60)
X_test_clip_scaled = scaler.transform(X_test_clip_raw) # Utilisez le scaler de la cellule 60
X_test_clip_pca = pca.transform(X_test_clip_scaled)    # Utilisez la pca de la cellule 60
X_test_pca_clip = pd.DataFrame(X_test_clip_pca, columns=[f'c_pc{i+1}' for i in range(X_test_clip_pca.shape[1])], index=X_test.index)

# 3. Transformation PCA du bloc 'vigg'
# Note: On utilise le scaler et la pca du train (cellule 61)
# Attention: Si vous avez écrasé les variables 'scaler' et 'pca', il faudra les renommer dans votre code train
X_test_vigg_scaled = scaler_vigg.transform(X_test_vigg_raw) 
X_test_vigg_pca = pca_vigg.transform(X_test_vigg_scaled)
X_test_pca_vigg = pd.DataFrame(X_test_vigg_pca, columns=[f'v_pc{i+1}' for i in range(X_test_vigg_pca.shape[1])], index=X_test.index)
```

```python
# Sélection des colonnes déjà filtrées pour chaque bloc
X_test_audio_filtered = X_test_audio_raw[X_train_audio_filtered.columns]
X_test_clip_filtered = X_test_pca_clip[X_train_clip_filtered.columns]
X_test_meta_filtered = X_test_meta_raw[X_train_meta_filtered.columns]
X_test_video_filtered = X_test_video_raw[X_train_video_filtered.columns]
X_test_vigg_filtered = X_test_pca_vigg[X_train_vigg_filtered.columns]

# Concaténation finale
X_test_filtered = pd.concat([
    X_test.iloc[:,0],
    X_test_audio_filtered,
    X_test_clip_filtered,
    X_test_meta_filtered,
    X_test_video_filtered,
    X_test_vigg_filtered
], axis=1)

print(f"Nombre de features X_test après : {X_test_filtered.shape[1]}")
# Vérification que les colonnes sont identiques au train
assert all(X_test_filtered.columns == X_train_filtered.columns)
```

```python
X_test_filtered.to_csv("../features/global_features/X_test_filtered.csv", index=False)
```
